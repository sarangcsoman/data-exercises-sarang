```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

```python
df = pd.read_csv("world_happiness_2023.csv")

print(f"Dataset: {len(df)} countries, {len(df.columns)} columns")
print("\nColumns:")
print(df.columns.tolist())

df.head()
```

```python
df.columns = [
    'Country',
    'Region',
    'Happiness_Score',
    'GDP',
    'Social_Support',
    'Life_Expectancy',
    'Freedom',
    'Generosity',
    'Corruption'
]

print(df.columns.tolist())
df[['Country', 'Region', 'Happiness_Score']].head()
```

```python
region_avg = (
    df.groupby('Region')['Happiness_Score']
    .mean()
    .reset_index()
    .sort_values('Happiness_Score', ascending=False)
)

print(region_avg)
```

```python
fig = px.bar(
    region_avg,
    x='Happiness_Score',
    y='Region',
    orientation='h',
    text=region_avg['Happiness_Score'].round(2),
    title='Western Europe Leads Regional Happiness, While South Asia Trails',
    labels={
        'Happiness_Score': 'Average Happiness Score',
        'Region': ''
    },
    color='Happiness_Score',
    color_continuous_scale='Blues'
)

fig.update_traces(
    textposition='outside',
    cliponaxis=False
)

fig.update_layout(
    template='plotly_white',
    title=dict(
        x=0.02,
        xanchor='left',
        font=dict(size=20)
    ),
    xaxis=dict(
        range=[0, 8],
        rangemode='tozero',
        showgrid=True,
        gridcolor='lightgrey',
        zeroline=True,
        zerolinewidth=2,
        zerolinecolor='black'
    ),
    yaxis=dict(
        categoryorder='total ascending'
    ),
    coloraxis_showscale=False,
    height=550,
    margin=dict(l=20, r=60, t=90, b=50)
)

fig.show()
```

```python
# Select the 8 happiest and 8 least happy countries
top8 = df.nlargest(8, 'Happiness_Score').copy()
top8['Group'] = 'Top 8'

bottom8 = df.nsmallest(8, 'Happiness_Score').copy()
bottom8['Group'] = 'Bottom 8'

# Combine the two groups and sort by happiness score
combined = pd.concat([bottom8, top8]).sort_values('Happiness_Score')

# Calculate the global average
global_avg = df['Happiness_Score'].mean()

print(f"Global average happiness score: {global_avg:.2f}")
print(combined[['Country', 'Happiness_Score', 'Group']])
```

```python
fig2 = px.bar(
    combined,
    x='Happiness_Score',
    y='Country',
    orientation='h',
    color='Group',
    color_discrete_map={
        'Top 8': '#167D5A',
        'Bottom 8': '#D95D39'
    },
    text=combined['Happiness_Score'].round(2),
    title='Happiness Scores Differ Sharply Between the Top and Bottom Countries',
    labels={
        'Happiness_Score': 'Happiness Score',
        'Country': '',
        'Group': 'Country Group'
    }
)

# Put scores outside the bars
fig2.update_traces(
    textposition='outside',
    cliponaxis=False
)

# Add global average reference line
fig2.add_vline(
    x=global_avg,
    line_dash='dash',
    line_color='#333333',
    line_width=2,
    annotation_text=f'Global average: {global_avg:.2f}',
    annotation_position='top'
)

fig2.update_layout(
    template='plotly_white',
    title=dict(
        x=0.02,
        xanchor='left',
        font=dict(size=20)
    ),
    xaxis=dict(
        range=[0, 9],
        rangemode='tozero',
        title='Happiness Score',
        showgrid=True,
        gridcolor='lightgrey',
        zeroline=True,
        zerolinewidth=2,
        zerolinecolor='black'
    ),
    yaxis=dict(
        categoryorder='array',
        categoryarray=combined['Country'].tolist()
    ),
    height=700
```

```python
fig2 = px.bar(
    combined,
    x='Happiness_Score',
    y='Country',
    orientation='h',
    color='Group',
    color_discrete_map={
        'Top 8': '#167D5A',
        'Bottom 8': '#D95D39'
    },
    text=combined['Happiness_Score'].round(2),
    title='The Happiest Countries Score Far Above the Least Happy',
    labels={
        'Happiness_Score': 'Happiness Score',
        'Country': '',
        'Group': 'Country Group'
    }
)

fig2.update_traces(
    textposition='outside',
    cliponaxis=False
)

fig2.add_vline(
    x=global_avg,
    line_dash='dash',
    line_color='#333333',
    line_width=2,
    annotation_text=f'Global average: {global_avg:.2f}',
    annotation_position='top'
)

fig2.update_layout(
    template='plotly_white',
    xaxis=dict(
        range=[0, 9],
        rangemode='tozero',
        title='Happiness Score',
        showgrid=True,
        gridcolor='lightgrey',
        zeroline=True,
        zerolinewidth=2,
        zerolinecolor='black'
    ),
    yaxis=dict(
        categoryorder='array',
        categoryarray=combined['Country'].tolist()
    ),
    height=700,
    margin=dict(l=20, r=70, t=110, b=50),
    legend=dict(
        title='Country Group',
        orientation='h',
        yanchor='bottom',
        y=1.01,
        xanchor='right',
        x=1
    )
)

fig2.show()
```
