```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

```python
df = pd.read_csv("co2_emissions.csv")

print(f"Loaded: {len(df)} rows")
print("Countries:", df["Country"].nunique())
print("Years:", df["Year"].min(), "-", df["Year"].max())

print("\nColumns:")
print(df.columns.tolist())

df.head()
```

```python
# Select the country to highlight
highlight_country = "China"

fig1 = go.Figure()

# Draw all other countries in light grey
for country in df["Country"].unique():
    country_df = df[df["Country"] == country].sort_values("Year")

    if country != highlight_country:
        fig1.add_trace(go.Scatter(
            x=country_df["Year"],
            y=country_df["CO2_per_capita"],
            mode="lines",
            line=dict(color="#EEEEEE", width=1.5),
            hovertemplate=(
                country + "<br>Year: %{x}"
                "<br>CO₂ per capita: %{y:.2f}<extra></extra>"
            ),
            showlegend=False
        ))

# Draw China in colour and label it directly at the end
china_df = df[df["Country"] == highlight_country].sort_values("Year")

fig1.add_trace(go.Scatter(
    x=china_df["Year"],
    y=china_df["CO2_per_capita"],
    mode="lines+markers",
    line=dict(color="#D95F02", width=3),
    marker=dict(size=5),
    name=highlight_country,
    showlegend=False,
    hovertemplate=(
        "China<br>Year: %{x}"
        "<br>CO₂ per capita: %{y:.2f}<extra></extra>"
    )
))

fig1.add_annotation(
    x=china_df["Year"].iloc[-1],
    y=china_df["CO2_per_capita"].iloc[-1],
    text="China",
    showarrow=False,
    xshift=25,
    font=dict(size=13, color="#D95F02")
)

fig1.update_layout(
    template="plotly_white",
    title=dict(
        text="China's CO₂ Emissions per Capita Have Risen Sharply Since 2000",
        x=0.02,
        xanchor="left"
    ),
    xaxis_title="Year",
    yaxis_title="CO₂ Emissions per Capita",
    xaxis=dict(showline=False, showgrid=False, zeroline=False),
    yaxis=dict(showline=False, showgrid=False, zeroline=False),
    height=550,
    margin=dict(l=60, r=100, t=90, b=50)
)

fig1.show()
```

```python
# Filter the Asian countries
asia_df = df[df["Region"] == "Asia"].copy()

print("Asian countries in dataset:")
print(asia_df["Country"].unique())

# Highlight China
highlight_country = "China"

fig2 = go.Figure()

# Draw all other Asian countries in grey
for country in asia_df["Country"].unique():
    country_df = asia_df[asia_df["Country"] == country].sort_values("Year")

    if country != highlight_country:
        fig2.add_trace(go.Scatter(
            x=country_df["Year"],
            y=country_df["CO2_Mt"],
            mode="lines",
            line=dict(color="#EEEEEE", width=1.5),
            showlegend=False,
            hovertemplate=(
                country + "<br>Year: %{x}"
                "<br>Total CO₂ emissions: %{y:.1f} Mt<extra></extra>"
            )
        ))

# Draw China in orange
china_asia = asia_df[
    asia_df["Country"] == highlight_country
].sort_values("Year")

fig2.add_trace(go.Scatter(
    x=china_asia["Year"],
    y=china_asia["CO2_Mt"],
    mode="lines+markers",
    line=dict(color="#D95F02", width=3),
    marker=dict(size=5),
    showlegend=False,
    hovertemplate=(
        "China<br>Year: %{x}"
        "<br>Total CO₂ emissions: %{y:.1f} Mt<extra></extra>"
    )
))

# Label China at the end of its line
fig2.add_annotation(
    x=china_asia["Year"].iloc[-1],
    y=china_asia["CO2_Mt"].iloc[-1],
    text="China",
    showarrow=False,
    xshift=25,
    font=dict(size=13, color="#D95F02")
)

fig2.update_layout(
    template="plotly_white",
    title=dict(
        text="China's Total CO₂ Emissions Have Risen Sharply Since 2000",
        x=0.02,
        xanchor="left"
    ),
    xaxis_title="Year",
    yaxis_title="Total CO₂ Emissions (Mt)",
    xaxis=dict(showline=False, showgrid=False, zeroline=False),
    yaxis=dict(showline=False, showgrid=False, zeroline=False),
    height=550,
    margin=dict(l=70, r=100, t=90, b=50)
)

fig2.show()
```

# Key Findings and Conclusion

## Task 1: CO₂ Emissions per Capita

China's CO₂ emissions per capita increased from approximately 1.3 in 2000 to around 4.4 in 2022. The overall trend shows a substantial increase over the period, although some year-to-year fluctuations occurred.

## Task 2: Total CO₂ Emissions in Asia

China's total CO₂ emissions rose from approximately 3,500 Mt in 2000 to over 12,000 Mt in 2022. The other Asian countries in the dataset had considerably lower total emissions. The chart highlights China's dominant contribution among the selected Asian countries.

## Overall Conclusion

Both charts show a substantial increase in China's CO₂ emissions between 2000 and 2022. The first chart examines emissions per capita, while the second examines total emissions. These measures provide different perspectives and should not be interpreted as interchangeable. The charts show trends in the dataset but do not establish the causes of those trends.
